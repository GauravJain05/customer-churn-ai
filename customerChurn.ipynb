**performing data analysis**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
import kagglehub
path = kagglehub.dataset_download("shantanudhakadd/bank-customer-churn-prediction")
csv_path = os.path.join(path,os.listdir(path)[0])
```

```python
data = pd.read_csv(csv_path)
```

```python
todrop = ["RowNumber","CustomerId","Surname"]
```

```python
data.drop(data[todrop],axis=1,inplace=True)
```

```python
data.head()
```

**feature engineering**

```python
data.info()
```

```python
cat_cols = ["Geography","Gender"]
```

```python
new_df = pd.get_dummies(data[cat_cols],drop_first=True,dtype=int)
```

```python
new_df.head()
```

```python
data.head()
```

```python
data.drop(data[cat_cols],axis=1,inplace=True)
```

```python
data.head()
```

```python
new_data = pd.concat([data,new_df],axis=1)
```

```python
new_data.head()
```

```python
new_data.info()
```

```python
new_data.shape
```

# Modeling

```python
x = new_data.drop(["Exited"],axis=1)
```

```python
new_data.head()
```

```python
y = new_data["Exited"]
```

```python
x.head()
```

```python
y.head()
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.10,random_state=42)
```

```python
print(x_train.shape,y_train.shape)
```

```python
print(x_test.shape,y_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rand_clf = RandomForestClassifier(n_estimators = 50)
```

```python
rand_clf.fit(x_train,y_train)
```

# Model Evaluation

```python
y_pred = rand_clf.predict(x_test)
```

```python
rand_clf.score(x_train,y_train)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
```

```python
confusion_matrix(y_test,y_pred)
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```
